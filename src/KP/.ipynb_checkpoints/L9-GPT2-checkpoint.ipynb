```python
num_return_sequences = 5
max_length = 30

model.eval()
model.to(device)

import tiktoken

enc = tiktoken.get_encoding('gpt2')
tokens = enc.encode('Hello, I\'m a language model,')
tokens = torch.tensor(tokens, dtype=torch.long)  # (8,)
tokens = tokens.unsqueeze(0).repeat(num_return_sequences, 1)  # (5, 8)
x = tokens.to(device)

# generate! right now x is (B, T), where B=5, T=8
torch.manual_seed(42)
torch.cuda.manual_seed(42)
while x.size(1) < max_length:
    # forward the model to get the logits
    with torch.no_grad():
        logits = model(x)  # (B, T, vocab_size)
        # take the logits at the last position
        logits = logits[:, -1, :]
        # get the probabilities
        probs = F.softmax(logits, dim=-1)
        # do top-k sampling of 50 (huggingface pipeline default)
        # topk_probs here becomes (5, 50), topk_indices is (5, 50)
        topk_probs, topk_indices = torch.topk(probs, 50, dim=-1)
        # select a token from the top-k probabilities
        ix = torch.multinomial(topk_probs, 1)  # (B, 1)
        # gather the corresponding indices
        xcol = torch.gather(topk_indices, -1, ix)  # (B, 1)
        # append to the sequence
        x = torch.cat((x, xcol), dim=1)

# print the generated text
for i in range(num_return_sequences):
      tokens = x[i, :max_length].tolist()
      decoded = enc.decode(tokens)
      print('>', decoded)
```

```python
##writing our own dataloader.

class DataLoaderLite:
    def __init__(self, B, T):
        self.B = B
        self.T = T
        
        # at init load tokens from disk and store them in memory
        with open('input.txt', 'r') as f:
            text = f.read ()
            
        enc = tiktoken.get_encoding ('gpt2')
        tokens = enc.encode (text)
        self.tokens = torch. tensor (tokens)
        
        print (f"loaded {len (self.tokens)} tokens")
        print(f"1 epoch = {len (self.tokens) // (B * T)} batches")
        
        # state
        self.current_position = 0
        
    def next_batch(self) :
        B, T = self.B, self.T
        buf = self.tokens[self.current_position: self.current_position+B*T+1]
        x = (buf [:-1]).view(B, T) # inputs
        y = (buf [1:]).view(B, T) # targets
        # advance the position in the tensor
        self. current_position+= B * T
        # if loading the next batch would be out of bounds, reset
        if self.current_position + (B * T + 1) > len(self.tokens):
            self. current_position = 0
        return x, y
```

```python
##using the data loader

train_loader = DataLoaderLite(B=4, T=32)

model = GPT(GPTConfig())
#model.to(device)

logits, loss = model(x, y)

optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4)

for i in range(51):
    x, y = train_loader.next_batch()
    #x, y = x.todevice(x, y)
    optimizer.zero_grad()
    logits, loss = model(x,y)
    loss.backward()
    optimizer.step()
    if i%10==0:
        print(f"step {i}, loss: {loss.item()}")
```

```python
#^^ this bit: NANOGPT_SCALE_INIT

%load_ext autoreload
%autoreload 2

from gpt_dev.train_gpt2 import GPT
import time

ddp_rank = 0
ddp_local_rank = 0
ddp_world_size = 1
master_process = True

device = 'cpu'
if torch.cuda.is_available():
  device = 'cuda'
elif hasattr(torch.backends, 'mps') and torch.backends.mps.is_available():
  device = 'mps'  # apple silicon
print(f'Using device {device}')

total_batch_size = 524288 # 2**19, ~0.5M, in number of tokens (like gpt3)
B = 8 #16 # micro batch size
T = 1024 # sequence length
assert total_batch_size & (B * T * ddp_world_size) == 0, "make sure total_batch_size is divisible by B * T"
grad_accum_steps = total_batch_size // (B * T * ddp_world_size)
print(f"total desired batch size: {total_batch_size}")
print(f"=> calculated gradient accumulation steps: {grad_accum_steps}")
train_loader = DataLoaderLite(B=B, T=T)

#train_loader = DataLoaderLite(B=8, T=1024) #too high? #B=16, T=1024) #more similar to desired workload
## B should max out the size of memory on your gpu.  

## if you have a high end gpu you can increase mat mul precision. 
torch.set_float32_matmul_precision("high")

model = GPT(GPTConfig(vocab_size=50304)) #nice numbers
model.to(device)
model = torch.compile(model)

max_lr = 3e-4
min_lr = max_lr * 0.1
warmup_steps = 1#3 #10
max_steps = 3#10 #50
def get_lr(it):
    """
    cosine decay learning rate with warmup (gpt3 paper)
    """
    # 1) linear warmup for warmup_iters steps
    if it < warmup_steps:
        return max_lr * (it+1) / warmup_steps
    # 2) if it › Ir_decay_iters, return min learning rate
    if it > max_steps:
        return min_lr
    # 3) in between, use cosine decay down to min learning rate
    decay_ratio = (it - warmup_steps) / (max_steps - warmup_steps)
    assert 0 <= decay_ratio <= 1
    coeff = 0.5 * (1.0 + math.cos(math.pi * decay_ratio)) # coeff starts at 1 and goes to 0
    return min_lr + coeff * (max_lr - min_lr)

#optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, betas=(0.9, 0.95), eps=1e-8)
optimizer = model.configure_optimizers(weight_decay=0.1, learning_rate=6e-4, device=device)

for i in range(max_steps): #51): #reduce for testing. got to step 50, loss: 5.87967586517334, dt: 2708.16ms
    t0 = time.time()
    optimizer.zero_grad()
    
    loss_accum = 0.0 # just to keep track of minibatch loss
    #this needs to repeat multiple times for the gradient accumilation (larger batch size). 
    for micro_step in range(grad_accum_steps):
        x, y = train_loader.next_batch()
        x, y = x.to(device), y.to(device)

        with torch.autocast(device_type=device, dtype=torch.bfloat16): 
            logits, loss = model(x,y) #bfloat16 (see the docs)'
        loss = loss / grad_accum_steps
        loss_accum += loss.detach()
        if micro_step % 20 == 0:
            print(f"micro step {micro_step} | loss: {loss.item():.4f}")
        loss.backward()
    
    norm = torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

    lr = get_lr(i)
    for param_group in optimizer.param_groups:
        param_group['lr'] = lr
    optimizer.step()
    
    #torch.cuda.synchronize() #finishes work on gpu before taking time.
    torch.mps.synchronize()
    t1 = time.time()
    dt = (t1-t0)*1000 # time difference in ms
    tokens_processed = train_loader.B * train_loader.T * grad_accum_steps
    tokens_per_sec = tokens_processed/dt
    print(f"step {i+1} | loss: {loss_accum.item():.6f} | norm: {norm:.4f} | dt: {dt:.2f}ms | tok/sec: {tokens_per_sec:.6f}")

#import sys; sys.exit(0
print("\n   ------------ Trained! Let's sample from it: ------------ \n")

for i in range(num_return_sequences):
      tokens = x[i, :max_length].tolist()
      decoded = enc.decode(tokens)
      print('>', decoded)
```

```python
##Flash attention - another kernel fusion operation. matmul. dropout, softmax. mask. matmul in 1 operation. 
### torch.compile can't find it. 
#flash attention is much more mindful about the memory organisation. (fewer reads and writes)
#NxN attention matrix (att line 46) is never read or written as a whole to HBM. 

#it is updated in a streaming fashion
#https://arxiv.org/pdf/2205.14135

#can be done on softmax also.
#https://arxiv.org/pdf/1805.02867

###also won't change mac speed (actually slows it slightly 2.9-3.1 s
```

```python
##but its tricky.. to do gradient accumulation (cant just loop batches before backward pass.

import torch
# super simple little MLP
net = torch.nn. Sequential(
    torch.nn.Linear(16, 32), 
    torch.nn.GELU(),
    torch.nn.Linear(32, 1), 
)
torch.random.manual_seed (42)
x = torch.rand(4, 16)
y = torch.rand(4, 1)
net.zero_grad()
yhat = net(x)
loss = torch.nn.functional.mse_loss(yhat, y)
loss.backward()
print (net[0].weight.grad.view(-1)[:10])
```

```python
#L = 1/N * sumN(y[i]-yhat[i])**2
## we need the normalization constant.

net.zero_grad() 
for i in range(4):
    yhat = net(x[i])
    loss = torch.nn. functional.mse_loss (yhat, y[i])
    loss = loss /4 # the normalizer
    loss.backward()
print (net[0].weight.grad.view(-1)[:10])

##matches 
```
